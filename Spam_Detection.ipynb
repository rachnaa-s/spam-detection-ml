```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

```python
import urllib.request
import zipfile
import os

url = "https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip"

urllib.request.urlretrieve(url, "sms_spam_collection.zip")

with zipfile.ZipFile("sms_spam_collection.zip", "r") as zip_ref:
    zip_ref.extractall(".")

print("Dataset downloaded successfully!")
```

```python
df = pd.read_csv(
    "SMSSpamCollection",
    sep="\t",
    header=None,
    names=["label", "message"]
)

df.head()
```

```python
print("Dataset shape:", df.shape)

print("\nLabels:")
print(df["label"].value_counts())
```

```python
df["label_num"] = df["label"].map({
    "ham": 0,
    "spam": 1
})

df.head()
```

```python
X = df["message"]
y = df["label_num"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=5000
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Training matrix shape:", X_train_tfidf.shape)
print("Testing matrix shape:", X_test_tfidf.shape)
```

```python
model = MultinomialNB()

model.fit(X_train_tfidf, y_train)

print("Model trained successfully!")
```

```python
y_pred = model.predict(X_test_tfidf)

print(y_pred[:20])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100)
```

```python
print(classification_report(
    y_test,
    y_pred,
    target_names=["Ham", "Spam"]
))
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Ham", "Spam"]
)

disp.plot()
plt.title("Spam Detection Confusion Matrix")
plt.show()
```

```python
def predict_spam(message):
    message_tfidf = vectorizer.transform([message])
    prediction = model.predict(message_tfidf)[0]

    if prediction == 1:
        return "SPAM"
    else:
        return "NOT SPAM"
```

```python
message = "Congratulations! You have won a free prize. Call now!"
print(predict_spam(message))
```

```python
message = "Hey, are we meeting at 6 pm today?"
print(predict_spam(message))
```

```python
def predict_spam_with_probability(message):
    message_tfidf = vectorizer.transform([message])

    prediction = model.predict(message_tfidf)[0]
    probabilities = model.predict_proba(message_tfidf)[0]

    spam_probability = probabilities[1] * 100

    if prediction == 1:
        result = "SPAM"
    else:
        result = "NOT SPAM"

    print("Message:", message)
    print("Prediction:", result)
    print(f"Spam Probability: {spam_probability:.2f}%")
```

```python
predict_spam_with_probability(
    "URGENT! You have won a free lottery prize. Claim now!"
)
```
